```python
def dfs(u,graph,vi):
    vi[u]=True
    print(u)
    for ng in graph[u]:
      if not vi[ng]:
        dfs(ng,graph,vi)
```

```python
# Example initialization for a BFS:
graph = {0: [1, 2], 1: [3, 4], 2: [], 3: [], 4: []}
visited = set()
start_node = 0
queue = [start_node]
visited.add(start_node)

while queue:
  w=queue.pop(0)
  print(w, end=" ")
  # graph(w) should be graph[w] to access the adjacency list
  for v in graph[w]:
    if v not in visited:
      visited.add(v) # Use add for sets
      queue.append(v)
```
