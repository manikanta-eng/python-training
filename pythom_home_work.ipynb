```python
class Node:

    def __init__(self, value):

        self.value = value

        self.link = None

class Queue:

    def __init__(self):

        self.start = None

        self.end = None

    def insert(self, value):

        new_node = Node(value)

        if self.end is None:

            self.start = self.end = new_node

            return

        self.end.link = new_node

        self.end = new_node

    def delete(self):

        if self.start is None:

            print("Underflow")

            return

        print("Deleted:", self.start.value)

        self.start = self.start.link

        if self.start is None:

            self.end = None

# Example

q = Queue()

q.insert(10)

q.insert(20)

q.insert(30)

q.delete()
```
