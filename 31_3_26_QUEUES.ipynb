```python
def is_valid_parenthesis(s: str) -> bool:
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}

    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():
            if not stack or mapping[char] != stack.pop():
                return False

    return not stack
print(f"{{([])}} --> {'valid' if is_valid_parenthesis("{([])}") else 'invalid'}")
print(f"() {{ }} [] --> {'valid' if is_valid_parenthesis("() { } []") else 'invalid'}")
print(f"[(]) --> {'valid' if is_valid_parenthesis("[(])") else 'invalid'}")
print(f"{{{{{{}}}}}} --> {'valid' if is_valid_parenthesis("{{{}}") else 'invalid'}") # Fixed: correctly escaped literal curly braces
print(f"}}{{ --> {'valid' if is_valid_parenthesis("}{") else 'invalid'}")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.size = 0

    def is_empty(self):
        return self.top is None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        self.size += 1
        print(f"Pushed: {data}")

    def pop(self):
        if self.is_empty():
            print("Stack is empty, cannot pop.")
            return None
        popped_data = self.top.data
        self.top = self.top.next
        self.size -= 1
        print(f"Popped: {popped_data}")
        return popped_data

    def peek(self):
        if self.is_empty():
            print("Stack is empty, no top element.")
            return None
        print(f"Top element: {self.top.data}")
        return self.top.data

    def get_size(self):
        print(f"Stack size: {self.size}")
        return self.size

    def display(self):
        if self.is_empty():
            print("Stack is empty.")
            return
        current = self.top
        elements = []
        while current:
            elements.append(str(current.data))
            current = current.next
        print("Stack (top to bottom): " + " -> ".join(elements))
my_stack = Stack()
my_stack.display()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.display()
my_stack.peek()
my_stack.get_size()
my_stack.pop()
my_stack.display()
my_stack.get_size()
my_stack.pop()
my_stack.pop()
my_stack.pop()
my_stack.peek()
my_stack.display()
```

**QUEUE**

```python
class Queue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, item):
        self.items.append(item)
        print(f"Enqueued: {item}")

    def dequeue(self):
        if self.is_empty():
            print("Queue is empty, cannot dequeue.")
            return None
        dequeued_item = self.items.pop(0)
        print(f"Dequeued: {dequeued_item}")
        return dequeued_item

    def peek(self):
        if self.is_empty():
            print("Queue is empty, no front element.")
            return None
        print(f"Front element: {self.items[0]}")
        return self.items[0]

    def size(self):
        current_size = len(self.items)
        print(f"Queue size: {current_size}")
        return current_size

    def display(self):
        if self.is_empty():
            print("Queue is empty.")
            return
        print("Queue (front to rear): " + " -> ".join(map(str, self.items)))


my_queue = Queue()
my_queue.display()


empty_list = []
my_queue.enqueue(empty_list)
my_queue.display()

my_queue.enqueue(10)
my_queue.enqueue("hello")
my_queue.display()
my_queue.peek()
my_queue.size()
my_queue.dequeue()
my_queue.display()
my_queue.dequeue()
my_queue.dequeue()
my_queue.dequeue()
my_queue.peek()
my_queue.display()
```

```python
class Queue:
    def __init__(self):
        self.queue = []


    def enqueue(self, item):
        self.queue.append(item)
        print(f"{item} inserted into queue")


    def dequeue(self):
        if self.is_empty():
            print("Queue is empty")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} removed from queue")


    def is_empty(self):
        return len(self.queue) == 0


    def display(self):
        print("Queue:", self.queue)


q = Queue()


q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()

q.dequeue()
q.display()

q.dequeue()
q.dequeue()
q.dequeue()
```

**QUEUE USING LIKED LIST**

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None



class Queue:
    def __init__(self):
        self.front = None
        self.rear = None


    def enqueue(self, item):
        new_node = Node(item)

        if self.rear is None:
            self.front = self.rear = new_node
            print(f"{item} inserted into queue")
            return

        self.rear.next = new_node
        self.rear = new_node
        print(f"{item} inserted into queue")


    def dequeue(self):
        if self.front is None:
            print("Queue is empty")
            return

        temp = self.front
        print(f"{temp.data} removed from queue")
        self.front = temp.next


        if self.front is None:
            self.rear = None


    def display(self):
        if self.front is None:
            print("Queue is empty")
            return

        temp = self.front
        print("Queue:", end=" ")
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")



q = Queue()


q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()

q.dequeue()
q.display()

q.dequeue()
q.dequeue()
q.dequeue()
```
