```python
# Plus symbol
n = 5

for i in range(n):
    for j in range(n):
        if i == n//2 or j == n//2:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 5 x 5 Hollow Rectangle (Border Only)

for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# X Symbol Pattern

n = 5

for i in range(n):
    for j in range(n):
        if i == j or i + j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Pattern Code

n = 5

for i in range(1, n + 1):
    for j in range(i):
        print((i + j) % 2, end="")
    print()
```

```python
H = int(input("Enter hours: "))

water = 0

for hour in range(1, H + 1):

    water += 50

    if hour % 2 == 0:
        water -= 10
    else:
        water -= 5

if water > 500:
    water -= 100
else:
    water += 50

print("Final water level =", water)
```

```python
L = int(input("Enter number of levels: "))

score = 10

for level in range(1, L + 1):

    score *= 2

    if level % 3 == 0:
        score += level * 5
    else:
        score -= level

    if score > 200:
        score -= 50

if score % 4 == 0:
    score = score // 4
else:
    score += 25

print("Final score =", score)
```
