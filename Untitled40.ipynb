```python
class stack:
  def __init__(self):
    self.stack=[]
  def push(self,data):

    self.stack.insert(0, data)
  def delete(self):

    if not self.is_empty():
      return self.stack.pop(0)
    else:
      print("Stack is empty, cannot delete.")
      return None
  def peek(self):

    if not self.is_empty():
      return self.stack[0]
    else:
      print("Stack is empty.")
      return None
  def is_empty(self):

    return len(self.stack) == 0
  def length(self):

    return len(self.stack)

s1=stack()
s1.push(10)
s1.push(20)
s1.push(30)

print("Current stack:", s1.stack)
print("Length:", s1.length())
print("Peek:", s1.peek())

s1.delete()
print("Stack after one deletion:", s1.stack)
print("Peek after deletion:", s1.peek())

s1.delete()
s1.delete()
s1.delete()
print("Stack after all deletions:", s1.stack)
print("Is empty:", s1.is_empty())
```
