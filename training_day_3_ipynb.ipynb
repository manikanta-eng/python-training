## recursion

```python
def display_list(lst, index=0):
    if index == len(lst):
        return
    print(lst[index])
    display_list(lst, index + 1)
numbers = [10, 20, 30, 40, 50]
display_list(numbers)
```

##reverse of a name

```python

def reverse_string(name):

    if len(name) == 0:
        return ""
    return reverse_string(name[1:]) + name[0]
name = input("Enter your name: ")
print("Original Name:", name)
print("Reversed Name:", reverse_string(name))
```

##palindrome

```python


def palindrome(word):
    if len(word) <= 1:
        return True
    if word[0] != word[-1]:
        return False
    return palindrome(word[1:-1])

text = input("Enter a word: ")
if palindrome(text):
    print(text, "is a Palindrome")
else:
    print(text, "is not a Palindrome")
```

#linear search

```python
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1
numbers = [10, 20, 30, 40, 50]
key = int(input("Enter element to search: "))
result = linear_search(numbers, key)
if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

##binary search

```python
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1

numbers = [10, 20, 30, 40, 50, 60, 70]

key = int(input("Enter element to search: "))

result = binary_search(numbers, key)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```
