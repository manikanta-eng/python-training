find radius of a circle when the area is 15.98
with this find perimeter of the circle using python code

```python
import math
area = 15.98
radius = math.sqrt(area / math.pi)
perimeter = 2 * math.pi * radius

print("Radius =", radius)
print("Perimeter =", perimeter)
```

```python
# Net Run Rate (NRR) Calculator

# taking inputs
runs_scored = int(input("Enter total runs scored: "))
overs_faced = float(input("Enter total overs faced: "))

runs_conceded = int(input("Enter total runs conceded: "))
overs_bowled = float(input("Enter total overs bowled: "))

# calculating run rates
run_rate_scored = runs_scored / overs_faced
run_rate_conceded = runs_conceded / overs_bowled

# calculating NRR
nrr = run_rate_scored - run_rate_conceded

# displaying results
print("Run Rate Scored =", round(run_rate_scored, 2))
print("Run Rate Conceded =", round(run_rate_conceded, 2))
print("Net Run Rate (NRR) =", round(nrr, 2))
```
