```python
def compute(x,func):
  return func(x)
def double(n):
  return n*2
def square(n):
  return n**2
result =compute(3,double)+compute(2,square)
print(result)
```

```python
def mystrey(a,b):
  if a%2==0:
    return b(a)
  else:
    return b(a)+1
def increment(x):
    return x+1
def decrement(x):
    return x-1
x=mystrey(5,increment)
y=mystrey(4,decrement)
print(x,y)
```
