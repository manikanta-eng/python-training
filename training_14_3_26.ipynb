```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
class DoublyLinkedList:
    def __init__(self):
        self.head=None
    def InsertionBeginning(self,new_data):
      new_node=Node(new_data)
      if self.head is None:
        self.head=new_node
      else:
        new_node.next=self.head
        self.head.prev=new_node
        self.head=new_node
    def PrintList(self):
      temp=self.head
      while temp:
        print(temp.data,end=" ")
        temp=temp.next
DLL=DoublyLinkedList()
DLL.InsertionBeginning(10)
DLL.InsertionBeginning(20)
DLL.InsertionBeginning(30)
DLL.Printlist()
```
